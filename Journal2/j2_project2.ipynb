```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def expFunc(a, b, x):
    return a*np.exp(-b*x)
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = expFunc(20, 2.5, x)
```

```python
sigma = 1.8 * (1-(x / (2*np.pi))) + 0.2
y = np.random.normal(y, sigma, 100)
```

```python
plt.errorbar(x, y, sigma*2, fmt='s', color='#a8e')
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
